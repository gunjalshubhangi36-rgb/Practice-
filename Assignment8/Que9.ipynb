9. Write a program to create a dictionary from two lists: one containing keys and the
other containing values.
- Keys = ['A', 'B', 'C']
- Values = [10, 20, 30]

```python
Keys = ['A', 'B', 'C']
Values = [10, 20, 30]

dict1 = {}

for i in range(len(Keys)):
    dict1[Keys[i]] = Values[i]

print(dict1)
```
