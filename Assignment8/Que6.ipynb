. Write a program to remove duplicate values from a dictionary.
- Input:
- {'a':10, 'b':20, 'c':10, 'd':30}
-
- Output:
- {'a':10, 'b':20, 'd':30}

```python
Dict1 = {'a':10, 'b':20, 'c':10, 'd':30}


res = set()
duplicate = {}

for key, value in Dict1.items():
    if value not in duplicate:
        res.add(value)
        duplicate[key] = value

print(res)
```
