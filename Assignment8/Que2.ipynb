2. Write a program to count the frequency of each element in a list using a dictionary.
- Input:
- [10,20,10,30,20,10]
-
- Output:
- {10:3, 20:2, 30:1}

```python
list1 = [3, 6, 7, 9, 3, 6]
print(list1)

dict1 = {}

for i in list1:
    if i in dict1:
        dict1[i] += 1
    else:
        dict1[i] = 1
print("Frequency of each element in a list :",dict1)
    
```
