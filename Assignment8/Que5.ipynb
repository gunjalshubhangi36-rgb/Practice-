5. Write a program to merge two dictionaries into a single dictionary.
- dict1 = {'a':1, 'b':2}
- dict2 = {'c':3, 'd':4}

```python
dict1 = {'a':1, 'b':2}
dict2 = {'c':3, 'd':4}

print("Dict 1 :",dict1)
print("Dict 2 :",dict2)

dict1.update(dict2)
print("Merge two dictionaries into a single dictionary :", dict1)
```
