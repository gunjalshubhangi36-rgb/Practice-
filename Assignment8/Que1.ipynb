1. Write a program to count the frequency of each character in a string using a
dictionary.
- Input:
- "programming"

- Output:
- {'p':1, 'r':2, 'o':1, ...}

```python
str1 = "programming"
print(str1)

dict1 = {}

for i in str1:
    if i in dict1:
        dict1[i] += 1
    else:
        dict1[i] = 1
print(dict1)
    
```
