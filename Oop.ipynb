Constructor

```python
class car:
    def __init__(self, color, name, price, milage):
        self.color = color
        self.name = name
        self.price = price
        self.milage = milage

    def display(self):
        print(self.name)
        print(self.color)
        print(self.price)
        print(self.milage)

obj = car("Black","BMW",100000,23)
obj.display()
```

```python
class car:
    def __init__(self, color, name, price, milage):   #self is used to store current value
        self.color = color
        self.name = name
        self.price = price
        self.milage = milage

    def display(self):
        print(self.name)
        print(self.color)
        print(self.price)
        print(self.milage)


color = input("Enter Color")
name = input("Enter Name")
price = input("Enter price")
milage = input("Enter Milage")

obj = car(color, name, price, milage)
obj.display()
```
