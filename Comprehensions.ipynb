Simple program 

```python
list1 = [1, 2, 3, 4, 5, 6, 7, 8]
list2 = []

for i in list1:
    if i % 2 == 0:
        list2.append(i*i)

print("Original list: ",list1)
print("Squre of number: ",list2)
```

Used comprehension in above list

```python
list1 = [1, 2, 3, 4, 5, 6, 7, 8]

#                 [ expresion   ,   iteration,   condition (optional)]
#exicution order  [   third     ,    first   ,       sencond         ]

list2 = [i**2 for i in list1 if i%2 == 0 ]

print(list1)
print(list2)
```

list comprhension

```python
list1 = [1,2,3,4,5]

list2= [i+i for i in list1 if i%2 == 0 ]

print(list1)
print(list2)
```

set comprhension

```python
set1 = {1, 2, 3, 4, 5,6}

set2 = {i for i in set1 if i%2 == 0}

print(set1)
print(set2)
```

Dictionary

```python
list1 = ["ram","sham","divya" ]

dict1 = {i: len(i) for i in list1 }

print(dict1)
```

```python
list1 = [1,2,3,4,5]

dict2 = {i: i*i for i in list1}

print(dict2)
```
