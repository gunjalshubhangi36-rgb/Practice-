Methods

```python
setOne = {"one", "two"}

setOne.add("Three")
print(setOne)

setOne.update({"Four", "Five"})
print(setOne)

setOne.remove("Four")           # display error when the other element is inter 
print(setOne)

setOne.discard("Three")         # can not display error when you can take other than set element
print(setOne)

setOne.pop()
print(setOne)

setOne.clear()
print(setOne)


```

```python
setOne = {"one", "two"}

temp = setOne.copy()
print(temp)

temp2 = setOne
print(temp2)

```

```python
setOne = {"one", "two"}
setTwo = {"four", "five", "one"}

print(setOne.union(setTwo))

print(setOne.intersection(setTwo))

print(setOne.difference(setTwo))
```

Built in function

```python
set1 = {"four", "five", "one"}

set2 = {1, 8, 5, 3}

print(len(set1))

print(max(set1))

print(min(set1))

print(sum(set2))

print(sorted(set2))

print(sorted(set1))

print(type(set1))

list1 = list(set1)
print(list1)





```

```python
set1 = {"four", "five", "one"}

for i, j in enumerate(set1):
    print(i, j)
```

```python
set1 = {"four", "five", "one", False}

print(any(set1))
print(all(set1))
```
