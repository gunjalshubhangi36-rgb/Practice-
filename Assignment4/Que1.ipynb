1. Write a program to count the number of vowels and consonants in a given string.


```python
str1 = input("Enter String")

a = str1.count("a")
e = str1.count("e")
i = str1.count("i")
o = str1.count("o")
u = str1.count("u")

str2 = a + e + i + o + u
print(str2)
```

```python
str1 = input("Enter String")

vowels = 0
consonants = 0

for i in str1:
    if i.isalpha():
        if i.lower() in 'aeiou':
            vowels += 1
        else:
            consonants +=1

print("Number of Vowels :", vowels)
print("Number of Consonant :", consonants)

```
