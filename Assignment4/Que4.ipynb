4. Write a program to count the occurrence of each character in a string.
Input : programming
Output:
p : 1
r : 2
o : 1

```python
str1 = input("Enter String")

p = str1.count("p")
print("p :", p)

r = str1.count("r")
print("r :", r)

o = str1.count("o")
print("o :", o)


```
