Tuple are immutable that can not be change or add value to it.

```python
tup1 = ("user1","user2","user3")

print(tup1)

print(tup1[1])       # to access element
```

Slicing

```python
tup1 = ("user1","user2","user3")

#[start, end, step]
print(tup1[0:2])
print(tup1[0:])
print(tup1[:3])
print(tup1[:])
print(tup1[::2])
print(tup1[::-1])        #used to reversed tuple

```

Etrable


```python
tup1 = ("user1","user2","user3")

for i, j  in enumerate(tup1):
    if i == 1:
        continue
    print(j)

```

```python
tup1 = ("user1","user2","user3","user4","user2")    #homogeneous

print(tup1)                               #allow duplicate

tup2 = ("user", 10 , 30, True)            #hetrogeneous

print(tup2)


```

Nested Tuple

```python
tup1 = ("user1","user2",("user4", "user5"),"user3")

print(len(tup1))
```

Positive Negative Indexing

```python
tup1 = ("user1","user2","user3")
#          0       1       2       Positive
#          -3      -2      -1      Negative

print(tup1[2])
print(tup1[-2])

```

Concatenation

```python
tup1 = ("user1","user2","user3")
tup2 = ("user4","user5","user6")

res = tup1 + tup2

print(res)
```

Methods Of Tuple

```python
tup1 = ("user1","user2","user3","user3")

print(tup1.count("user3"))     # count the value occurs in the tuple

print(tup1.index("user1"))     # return index

```

Built in Function

```python
tup1 = ("user1","user2","user3","user3")

print(len(tup1))

print(max(tup1))

print(min(tup1))
```

```python
tup1 = (2, 4, 5, 7)

print(sum(tup1))
```

```python
tup1 = (2, 4, 5, 7)

temp = 0

for i in tup1:
    temp += 1

print(temp)

print(sorted(tup1))
```

```python
tup1 = (2, 4, 5, 7)

print(type(tup1))

templist = list(tup1)
print(templist)

set1 = set(tup1)
print(set1)

str1 = str(tup1)
print(str1)

```

```python
tup1 = (2, 4, 5, 7)

for i, j in enumerate(tup1):
    print(i, j)
```

```python
tup1 = ("user1","user2",False,"user3")

print(any(tup1))

print(all(tup1))

```
