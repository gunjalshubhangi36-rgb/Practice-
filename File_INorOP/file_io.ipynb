Write Mode


```python
str1 = "hello word"
str2 = "hello user 2"

fobj = open("demo.text", "w")     # in this the data only write and new data is added then existing are deleted and new added
fobj.write(str2)

fobj.close()
```

Read Mode

```python
str1 = "hello word"
str2 = "hello user 2"

fobj = open("demo.text", "a")   # it can be used to append or add data in the file
fobj.write(str2)

fobj.close()
```

Append Mode

```python
fobj = open("demo.text", "r")
content = fobj.read()
print(content)

fobj.close()
```

Functions

write()

```python
str1 = "hello word"
str2 = "hello user 2"

fobj = open("demo.text", "w")     # in this the data only write and new data is added then existing are deleted and new added
fobj.write(str2)

fobj.close()
```

writelines()

```python
list1 = ["username\n","usertwo\n","user3\n"]

fobj = open("temp5.txt","w")

fobj.writelines(list1)

fobj.close()
```

read()

```python
fobj = open("temp5.txt", "r")

info = fobj.read()
print(info)

fobj.close()
```

readline()

```python
fobj = open("temp5.txt", "r")

info = fobj.readline()
print(info)

info = fobj.readlines()
print(info)

fobj.close()
```

with keyword

```python
with open("temp5.txt","r") as fobj:
    info = fobj.read()
    print(info)

fobj.close()
```
