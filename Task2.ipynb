2. Write a Python program using nested loops to find and print all prime numbers
between 1 and 100. 

```python
for i in range(1, 101):

    if i > 1:

        for j in range(2, i):
            if (i % j) == 0:
                break
        else:
            print(i, end=" ")
        
```

```python
"""6. Write a Python program to find the second largest unique number from a list without
using sort(). 
"""

lisst1 = [10, 20, 30, 40, 50, 50, 40, 30]

list2 = list(set(lisst1))
print("Unique numbers:", list2)

largest = max(list2)
list2.remove(largest)

print("Second largest:", max(list2))
    
```

9. Write a Python program to count the frequency of each word in a sentence and store
the result in a dictionary. 

```python
sentence = "This is a sample sentence for testing the word count program."
print("Original sentence:", sentence)

dict1 = {}

for i in sentence.split():
    if i in dict1:
        dict1[i] += 1
    else:
        dict1[i] = 1
print("Word count:", dict1)
```

10. Given two dictionaries containing product names and prices, merge them into one
dictionary. If a product exists in both, keep the higher price. 

```python
dict1 = {"product1": 10, "product2": 20, "product3": 30}
dict2 = {"product3": 5, "product4": 15, "product5": 30}

dict = dict1.copy()

for key, value in dict2.items():
    if key in dict:
        dict[key] = max(dict[key], value)
    else:
        dict[key] = value
print(dict)
```

11.Create a function calculator(a, b, operation) that performs addition, subtraction,
multiplication, and division based on the given operation. 

```python
def calculator(a, b, operation):
    if operation == "add":
        return a + b
    elif operation == "subtract":
        return a - b
    elif operation == "multiply":
        return a * b
    elif operation == "divide":
        if b != 0:
            return a / b
        else:
            return "Error: Division by zero"
    else:
        return "Invalid operation"

calculator(10, 5, "add")
```

12.Create a function that accepts a list of numbers and returns the largest number,
smallest number, sum, and average. 

```python
def calculate(numbers):
    largest = max(numbers)
    smallest = min(numbers)
    total = sum(numbers)
    average = total / len(numbers)

    return largest, smallest, total, average


numbers = [10, 20, 5, 30, 15]

largest, smallest, total, average = calculate(numbers)

print("Largest number:", largest)
print("Smallest number:", smallest)
print("sum of numbers:", total)
print("Average of numbers:", average)
```

13.Create a BankAccount class with account holder, account number, and balance.
Add methods for deposit, withdrawal, and displaying balance. Prevent withdrawal
when the balance is insufficient. 

```python
class BankAccount:
    def __init__(self, account_holder, account_number, balance):
        self.account_holder = account_holder
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdrawal(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance!")

    def display_balance(self):
        print("Account Holder:", self.account_holder)
        print("Account Number:", self.account_number)
        print("Balance:", self.balance)



account = BankAccount("Shubhangi", "1234567890", 10000)

account.display_balance()

account.deposit(5000)
account.withdrawal(3000)
account.display_balance()


account.withdrawal(20000)
```

15.Create a Vehicle class with brand and speed. Create Car and Bike child classes with
additional attributes. Use inheritance and method overriding to display their
complete details. 

```python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
    def display(self):
        print("Brand:", self.brand)
        print("Speed:", self.speed)

class Bike(Vehicle):
    def __init__(self, brand, speed, type):
        super().__init__(brand, speed)
        self.type = type

    def display(self):
        super().display()
        print("Type:", self.type)

class Car(Vehicle):
    def __init__(self, brand, speed, model):
        super().__init__(brand, speed)
        self.model = model

    def display(self):
        super().display()
        print("Model:", self.model)

obj1 = Bike("Yamaha", 120, "Sports")
obj1.display()

obj2 = Car("Toyota", 180, "Sedan")
obj2.display()
```
