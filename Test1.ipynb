Write a program to find the sum of largest and second-largest number from a list .

```python
list1 = [6,7,3,4,9]

sorted_list = sorted(list1)

largestNo = sorted_list[-1]
secondLN = sorted_list[-2]

add = largestNo + secondLN

print("The sum of largest and second largest number of list is:",add)

```

Write a program to accept a string and count the frequency of each character using a dictionary. Ignore spaces. 

```python
str1 = input("Enter string")
str1.lower().strip()

res = {}

for i in str1:
    if i in res:
        res[i] += 1
    else:
        res[i] = 1
print("Frequency: ", res, end=" ")
print(" ")


```

Write a function that accepts a list of numbers and returns a another list containing: 
o	Sum of all numbers 
o	Number of even elements 
o	Number of odd elements 
o	Largest number 

```python
def sum_fun(*args):
    add = sum(args)

    even = list(filter(lambda i: i%2==0, args))
    odd  = list(filter(lambda i: i%2!=0, args))

    largestNo = max(args)

    print("addition of all numbers: ",add)
    print("even number in the list: ", even)
    print("odd number in the list: ", odd)
    print("largest number in the list: ", largestNo)

sum_fun(23,45,55,78,22,34)
```
