List Methods

```python
std = ["user1","user2","user3"]

std.append("user4")         #add element in the list
print(std)

std.extend(["user5","user6"]) #add multiple value at a time
print(std)

std.insert(2,"user4")          #value add in list any ware
print(std)

std.remove("user3")            #for remove specific element
print(std)

std.index("user1")             #used to display index of the value           
print(std)

print(std.count("user4"))     # number of time value occurs in the list


#std.clear()
#print(std)



```

```python
listOne = ["user1","user2"]

print("original :",listOne)
print(id(listOne))

copyone = listOne.copy()      # store copy of array
print("copy1 :",copyone)      #shallow copy
print(id(copyone))


copyTwo = listOne
print("copy2 :", copyTwo)       #deep copy
print(id(copyTwo))
print()

copyTwo.append("user3")
print("original :",listOne)
print("copy1 :",copyone)      #shallow copy
print("copy2 :", copyTwo)       #deep copy


```
