While Loop

```python
#Synatax = While Condition:
#             Block

i = 1

while i <= 5:
    print("Hello")

    i += 1
```

For Loop

```python
list1 = [1, 2, 3, 4, 5]

for i in list1:
    print(i)
```

Rang Funaction

```python
for i in range(1, 11):
    print(i * 5)
```

Brake Statment

```python
for i in range(1, 11):
    if i == 5:
        break
    print(i)
```

Continue Statment

```python
for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```

Odd Number 

```python
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i)
```

Pass Statement

```python
for i in range(1, 11):
    if i == 5:
        pass                      #used when you have to stay empty in the block of code
    print(i)
```

Nested for

```python
for i in range(1, 3):
    for j in range(1, 6):
        print(i, j, end="   ")

    print("")
    print("")
```

```python
col = 3
row = 1

for i in range(col):
    for j in range(row):
        print("*", end=" ")

    print("")
    row +=1
    
```
