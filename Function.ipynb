1. positional argument

```python
def student(name, roll):
    print("Student Name:", name, "Roll No:", roll)

student(101, "shubhangi")
student("shubhangi", 101)
```

```python
def sub(num1, num2):
    
    print(num1 - num2)

sub(10, 20)
```

2. keyword arguments

```python
def student(name, roll_no):
    print("Student Name:", name, "Roll No:", roll_no)

student(name="shubhangi", roll_no=50)
student(roll_no=50, name="shubhangi")
```

3. Default argument

```python
def student(name="temp",roll=0):
    print("Student name is", name, "Roll No is", roll)
student()
student(name="sam",roll=40)
```

4. Variable length arguments

```python
def temp(*args):
    for i in args:
        print(i)

temp(10, 20, 30, 40, 50)
```

5. Keyword

```python
def details(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

    print(kwargs["name"])             #used to access the value of a specific key in the dictionary
    print(kwargs["roll"])

details(name="shubhangi", roll=101, age=20, city="pune")
```

Functional Programming

1. Lambada Funation


```python
def squre(num1):
    print(num1*num1)
squre(10)
```

```python
squre = lambda num1: num1*num1
print(squre(10))
```

```python
add = lambda num1, num2: num1 + num2
print(add(10, 20))
print(add(40, 20))
print(add(60, 20))
```

```python
listOne = [10, 20, 30, 40, 50]

#1. map - specific opraction
#2. filter - specific condition
#3. reduce - specific operation on the entire list(sum)
```

```python
listOne = [10, 20, 30, 40, 50]
listTwo = []

for i in listOne:
    listTwo.append(i*i)
print(listOne)
print(listTwo)
```

1.Map

```python
listOne = [10, 20, 30, 40, 50]

listTwo = list(map(lambda num: num*num, listOne))
print(listTwo)
```

2. Filter

```python
listOne = [10, 20, 30, 40, 50]

even = list(filter(lambda num: num%2==0, listOne))

print(even)
```

3. Reduce

```python
from functools import reduce

listOne = [10, 20, 30, 40, 50]

add = reduce(lambda num1, num2: num1 + num2, listOne)

print(add)

print(sum(listOne))
```
