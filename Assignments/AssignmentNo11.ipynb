1. Write a program to handle ZeroDivisionError when dividing two numbers entered by
the user. 

```python
num1 = int(input("Enter number 1"))
num2 = int(input("Enter number 2"))

try:
    res = num1/num2
    print(res)
except ZeroDivisionError as err:
    print("Error: number is not divide by zero :", err)

else:
    ("Divide succesfully")
```

3. Write a program to handle IndexError while accessing an element from a list. 


```python
list1 = [1, 2, 3, 4]

try:
    print(list1[5])
except IndexError as err:
    print("Error: Index out of bound", err)
else:
    ("Successfull")
```
