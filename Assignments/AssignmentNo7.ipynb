1. Write a program to find the union of two sets without using the union() method.

```python
set1 = {2, 3, 4, 7, 9 ,4 ,6}
set2 = {2, 8, 6, 0, 1, 3, 7}

union = set1

for i in set2:
    if i not in union:
        union.add(i)

print("Union: ", union)
```

2. Write a program to remove all duplicate elements from a list using a set.

```python
list1 = [2, 3, 4, 2, 9 ,3 ,6]
print("Original list :", list1)
unique = list(set(list1))
print("After removing duplicate :", unique)
```

3. Write a program to find all common elements between two lists using sets.

```python
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]

common = set(list1) & set(list2)

print("Common elements:", common)
```

4. Write a program to count the number of unique elements in a list using a set.

```python
list1 = [2, 3, 4, 7, 9 ,4 ,6]


union = set(list1)

print("Unique element : ", union)
print("Count of unique element: ", len(union))

```

6. Write a program to create a new set containing only the even numbers from a given
set.

```python
set1 = {7, 9, 6, 4, 8}
print("Set 1: ", set1)

res = set()

for i in set1:
    if i % 2 == 0:
        res.add(i)
print("Set containing only even numbers :", res)
```

7. Write a program to merge three sets into a single set containing only unique
elements.

```python
set1 = {2, 7, 9, 5}
set2 = {6, 8, 0, 1}
set3 = {7, 8, 5, 3}

newset = set1.copy()

for i in set2:
    newset.add(i)

for i in set3:
    newset.add(i)

print(newset)
```

9. Given multiple lists, find all elements that appear in every list using sets.

```python
list1 = [1, 2, 7, 8]
list2 = [5, 6, 7, 8]
list3 = [4, 7, 3, 8]

set1 = set(list1)
set2 = set(list2)
set3 = set(list3)

print(set1 & set2 & set3)
```
