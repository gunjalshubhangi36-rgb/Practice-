1. Write a program to create a text file and store the name, age, and marks of a student
in it. 

```python
name = input("Enter name: ")
age = int(input("Enter age: "))
marks = int(input("Enter marks: "))

fobj = open("temp1.txt", "w")

fobj.write(f"Name of student: {name}\n")
fobj.write(f"Age of student: {age}\n")
fobj.write(f"Marks of student: {marks}\n")

fobj.close()

print("Data written successfully")
```

6. Write a program to read a file and create another file containing only the lines that
contain the word "Python". 

```python
with open("demo1.txt","r") as file1:
    data = file1.readlines()

    with open("temp2.txt","w") as file2:
        for lines in data:
            if "python" in lines.lower():
                file2.write(lines)

    file2.close()
file1.close()
```

8. Write a program to read a file and find the longest word present in it. 


```python
with open("demo1.txt","r") as file:
    data = file.read()

word = data.split()
longest = word[0]

for i in word:
    if (len(i)) > (len(longest)):
        longest = i

print("The longest word in file: ", longest)
```

9. Write a program to read a file containing numbers and calculate their sum, average,
maximum, and minimum.

```python
with open("demo2.txt","r") as file:
    data = file.read()
    print("data: ",data)

nums = list(map(int, data.split()))    
    
adds = sum(nums)
avg = adds/len(nums)
maxm = max(nums)
minm = min(nums)
print(f"Addition: {adds}\nAverage: {avg}\nMaximum: {maxm}\nMinimum: {minm}")
```
