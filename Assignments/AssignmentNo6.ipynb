1. Write a program to find the largest and smallest element in a tuple without using
max() and min().

```python
tup1 = (2, 3, 4, 7, 9 ,4 ,6)

smallest = tup1[0]
largest = tup1[0]

for i in tup1:
    if i < smallest:
        smallest = i
    if i > largest:
        largest = i
print("Smallest :",smallest)
print("Largest :",largest)

```

2. Write a program to count the number of even and odd numbers in a tuple.

```python
tup1 = (2, 3, 4, 7, 9 ,4 ,6)

even = 0
odd = 0

for i in tup1:
    if i % 2 == 0:
        even += 1
    else:
        odd += 1

print("Count of the even numbers :", even)
print("Count of the odd numbers :", odd)
```

3. Write a program to find all common elements between two tuples.

```python
tup1 = (2, 3, 4, 7, 9 ,4 ,6)
tup2 = (2, 8, 6, 0, 1, 3, 7)

print("Common element in the tuple :")

for i in set(tup1):
    for j in set(tup2):
       if i == j :
          print(i)
```

4. Write a program to create a new tuple containing only positive numbers from a
given tuple.


```python
tup1 = (2, 3, -4, 7, 9 ,-4 ,-6)

tup2 = ()
for i in tup1:
    if i > 0:
        tup2 += (i, )
print("Positive numbers in the tuple :",tup2)
```

5. Write a program to remove duplicate elements from a tuple while preserving the
original order.

```python
tup1 = (2, 3, 4, 2, 9 ,3 ,6)
print("Original Tuple :", tup1)

new = ()
for i in tup1:
    if i not in new:
        new += (i, )
print("After removing duplicate :", new)
```

7. Write a program to find the intersection of two tuples.

```python
tup1 = (2, 3, 4, 7, 9 ,4 ,6)
tup2 = (2, 8, 6, 0, 1, 3, 7)

print(" Intersection:", end = " ")

for i in set(tup1):
    for j in set(tup2):
       if i == j :
          print(i, end = " ")
```

8. Write a program to find the union of two tuples (unique elements only).

```python
tup1 = (2, 3, 4, 7, 9 ,4 ,6)
tup2 = (2, 8, 6, 0, 1, 3, 7)

union = tup1

for i in tup2:
    if i not in union:
        union += (i, )

print("Union: ", union)

```
