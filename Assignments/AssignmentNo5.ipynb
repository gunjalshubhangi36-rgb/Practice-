1. Write a program to count the number of even and odd numbers in a list.


```python
list = [2,4,6,7]

even = 0
odd = 0

for i in list:
    if i % 2 == 0:
        even += 1
    else:
        odd += 1
print("count of even number :", even)
print("count of odd number :", odd)
```

2. Write a program to create a new list containing only the positive numbers from a given
list.
- Input:
- [-10, 20, -5, 40, 0, 15]
-
- Output:
- [20, 40, 15]

```python
listOne = [-10, 20, -5, 40, 0, 15]
res = []
for i in listOne:
    if i > 0:
        res.append(i)

print("list of positive numbers :" ,res)
        
```

3. Write a program to remove all duplicate elements from a list while preserving the
original order.
- Input:
- [10,20,30,20,40,10,50]
-
- Output:
- [10,20,30,40,50]


```python

listOne = [10,20,30,20,40,10,50]
list=[]
for i in listOne:
    if i not in list:
        list.append(i)

print("Output :",list)
```

8. Write a program to check whether a list is a Palindrome.
- Input:
- [1,2,3,2,1]
-
- Output:
- Palindrome

```python
listOne = [1,2,3,2,1]

listTwo = list(reversed(listOne))

if listOne == listTwo:
    print("list is palindrom")
else:
    print("list is not palindrom")
```

9. Write a program to find all common elements between two lists.
- List1 = [10,20,30,40]
- List2 = [20,40,60]
-
- Output:
- [20,40]

```python
List1 = [10,20,30,40]
List2 = [20,40,60]

comman = []
for i in List1:
    for j in List2:
        if i == j:
            comman.append(i)

print("Comman element in two lists :", comman)
```

10. Write a program to separate all even and odd numbers into two different lists.
- Input:
- [10,15,20,25,30]
-
- Output:
- Even = [10,20,30]
- Odd = [15,25]

```python
listOne = [10,15,20,25,30]
even = []
odd = []

for i in listOne:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)
print("Even :", even)
print("Odd :", odd)
```
