1. Write a program to count the frequency of each character in a string using a
dictionary.
- Input:
- "programming"

- Output:
- {'p':1, 'r':2, 'o':1, ...}

```python
str1 = "programming"
print(str1)

dict1 = {}

for i in str1:
    if i in dict1:
        dict1[i] += 1
    else:
        dict1[i] = 1
print(dict1)
    
```

6. Write a program to remove duplicate values from a dictionary.
- Input:
- {'a':10, 'b':20, 'c':10, 'd':30}
-
- Output:
- {'a':10, 'b':20, 'd':30}

```python
Dict1 = {'a':10, 'b':20, 'c':10, 'd':30}


res = set()
duplicate = {}

for key, value in Dict1.items():
    if value not in duplicate:
        res.add(value)
        duplicate[key] = value

print(res)
```

9. Write a program to create a dictionary from two lists: one containing keys and the
other containing values.
- Keys = ['A', 'B', 'C']
- Values = [10, 20, 30]

```python
Keys = ['A', 'B', 'C']
Values = [10, 20, 30]

dict1 = {}

for i in range(len(Keys)):
    dict1[Keys[i]] = Values[i]

print(dict1)
```

10. Write a program to create a student result dictionary containing name and marks of
5 students, then display the topper.

```python
students = {'Amit': 85, 'Neha': 92, 'Ravi': 78, 'Sneha': 95, 'Kiran': 88 }

topper = max(students, key=students.get)

print('Topper :', topper)
print('Marks :', students[topper])
```
