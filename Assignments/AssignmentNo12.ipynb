2. What will be the output? Explain the LEGB rule step by step.
x = "Global"
def outer():
x = "Enclosing"
def inner():
print(x)
inner()
outer()

```python
x = "Global"
def outer():
    x = "Enclosing"
    def inner():
        print(x)      #Enclosing
    inner()
outer()
```

3. Identify the Local, Enclosing, Global, and Built-in names in the following program:
x = 10
def outer():
y = 20
def inner():
z = 30
print(x, y, z, len("Python"))
inner()
outer()

```python
x = 10          #global
def outer():
    y = 20      #enclosing
    def inner():
        z = 30      #local
        print(x, y, z, len("Python"))
    inner()             #len("Python") Built-in
outer()
```

```python
x = 100
def outer():
    x = 200
    def inner():
        x = 300     #shadowed
        print(x)

    inner()
    print(x)

outer()
print(x)
```
