1. Print numbers from 1 to 100 using a for loop.


```python
for i in range(1, 101):
    print(i)
```

2. Print numbers from 100 to 1 using a while loop.


```python

num = 100

while num >= 1:
    print(num)
    num -= 1
```

3. Print all even numbers between 1 and 100.

```python
for i in range(2, 101):
    if i % 2 == 0:
        print(i)
```

4. Print all odd numbers between 1 and 100.


```python
for i in range(1, 100):
    if i % 2 != 0:
        print(i)
```

5. Print the multiplication table of any number entered by the user.


```python
num = int(input("Enter a number: "))
for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```
