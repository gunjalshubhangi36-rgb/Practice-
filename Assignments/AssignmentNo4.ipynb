2. Write a program to check whether a given string is a Palindrome.
Input: madam
Output : Palindrome

```python
str1 = input("Enter String")

str2 = reversed(str1)

if str1 == str2:
    print("String is palindrom")
else:
    print("String is not palindrom")
```

3. Write a program to reverse a string without using slicing ([::-1]) or built-in functions.

```python
string = input("Enter a string: ")

reverse = ""

for i in string:
    reverse = i + reverse

print("Reversed string:", reverse)
```

5. Write a program to remove all duplicate characters from a string while maintaining their original order.
Input : programming
Output : progamin

```python
string = input("Enter a string: ")

result = ""

for i in string:
    if i not in result:
        result = result + i

print("Output:", result)
```

6. Write a program to find the longest word in a sentence.
Input: Python is an amazing programming language
Output: programming

```python
sentence = input("Enter a sentence: ")

words = sentence.split()

longest = words[0]

for word in words:
    if len(word) > len(longest):
        longest = word

print("Longest word:", longest)
```

7. Write a program to capitalize the first letter of every word without using the title() method.
Input: welcome to python
Output: Welcome To Python

```python
Str1 = input("Enter string")

words = Str1.split()


for word in words:
    print(word.capitalize(), end = " ")
```
