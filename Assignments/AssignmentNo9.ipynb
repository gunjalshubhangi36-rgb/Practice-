1. Write a function that accepts two numbers and returns the sum, difference, using
multiple return values. 

```python
def res(num1, num2):
    add = num1 + num2
    sub = num1 - num2
    return add, sub
print(res(10, 5))
```

2. Write a function that accepts a list of numbers and returns the largest and smallest
number.

```python
def nums(*args):
    samllest = min(args)
    largest = max(args)
    return samllest, largest
print(nums(10, 20, 30, 40, 50))
```

3. Write a function using *args to calculate the sum, average, maximum, and
minimum of any number of integers. 

```python
def nums(*args):
    add = sum(args)
    avg = add / len(args)
    max1 = max(args)
    min1 = min(args)
    return add, avg, max1, min1

print(nums(10, 20, 30, 40, 50))
```

4. Write a function using **kwargs to display employee details (id, name, department,
salary) in a formatted manner. 

```python
def emp(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

emp(id = 101, name = "sam", department= "IT", salary = 50000)
```

5. Write a function that accepts a string and returns the number of uppercase letters,
lowercase letters, digits. 

```python
def result(str):
    upper = 0
    lower = 0
    digit = 0
    for i in str:
        if i.isupper():
            upper += 1
        elif i.islower():
            lower += 1
        elif i.isdigit():
            digit += 1
    return upper, lower, digit
result("Hello World 123")
```

6. Write a function that checks whether a given string is a Palindrome. 


```python
def palindrome(str):
    str1 = str.lower()
    if str1 == str1[::-1]:
        print("The string is a palindrome")
    else:
        print("The string is not a palindrome")
    
palindrome("Racecar")
```

8. Create a menu-driven calculator using functions that performs Addition,
Subtraction, Multiplication, Division, and Exit. 

```python

def add(num1, num2):
    return num1 + num2
def sub(num1, num2):
    return num1 - num2
def mul(num1, num2):
    return num1 * num2
def div(num1, num2):
    if num2 != 0:
        return num1 / num2
    else:
        return "Error: Division by zero"
while True:
    print("Menu:")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Exit")
    
    choice = input("Enter your choice (1-5): ")
    
    if choice == '5':
        print("Exiting the calculator.")
        break
    
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    
    if choice == '1':
        result = add(num1, num2)
        print("Result", result)
    elif choice == '2':
        result = sub(num1, num2)
        print("Result", result)
    elif choice == '3':
        result = mul(num1, num2)
        print("Result", result)
    elif choice == '4':
        result = div(num1, num2)
        print("Result", result)
    else:
        print("Invalid choice. Please try again.")
    
```

9. Write a function that accepts a dictionary of students and their marks, then returns
the name of the topper and the highest marks. 

```python
def topper(**kwargs):
    max_marks = 0
    topper_name = ""
    
    for name, marks in kwargs.items():
        if marks > max_marks:
            max_marks = marks
            topper_name = name
            
    return topper_name, max_marks
topper(divya=85, sanu=92, sam=78, shubhangi=95, dipali=88)
    
```

10. Write a function that accepts a sentence and returns the frequency of each word
using a dictionary. 

```python
def fre(str):

    str2 = str.lower().split()
    res =" "

    for i in str2:
        if i not in res:
            print(i, ":", str2.count(i))
            res += i + " "

fre("This is a sample string. This string is for testing the frequency.")
```

11. Write a function that accepts a list of integers and returns:
• Sum of all numbers
• Count of even numbers
• Count of odd numbers

```python
def calculate(nums):
    add = sum(nums)
    evanNo = len(list(filter(lambda num: num%2==0, nums)))
    oddNo = len(list(filter(lambda num: num%2!=0, nums)))
    return add, evanNo, oddNo
calculate([56, 87, 87, 90, 45, 23, 12])
```
