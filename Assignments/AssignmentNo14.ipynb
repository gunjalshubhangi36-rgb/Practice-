1. Student Information System
Create a Student class with:
• name
• roll_no
• branch
• marks
Create 3 student objects and create a method display() to print their details.


```python
class student:
    def __init__(self, name, roll_no, branch, marks):
        self.name = name
        self.roll_no = roll_no
        self.branch = branch
        self.marks = marks

    def display(self):
        print(self.name, self.roll_no, self.branch, self.marks)


obj1 = student("Shubhangi",10,"CS",95)
obj2 = student("Divya",11,"CS",90)
obj3 = student("Dipali",12,"CS",85)
obj1.display()
obj2.display()
obj3.display()
```

2. Employee Salary
Create an Employee class with:
• name
• employee_id
• salary
Create a method display_salary() that displays the employee's salary.
Create 3 employee objects.
: Don't assign the values directly inside the class. Pass them while creating objects

```python
class Employee:
    def __init__(self, name, employee_id, salary):
        self.name = name
        self.employee_id = employee_id
        self.salary = salary

    def display_salary(self):
        print(self.name, self.employee_id, self.salary)

obj1 = Employee("Shwetali", 101, 100000)
obj2 = Employee("Saniya", 102, 50000)
obj3 = Employee("Divya", 103, 15000)

obj1.display_salary()
obj2.display_salary()
obj3.display_salary()
```

3. Mobile Phone
Create a Mobile class with:
• brand
• model
• price
Create a method:
display_details()
Create 3 mobile objects with different information.
Expected concept: Constructor + objects.

```python
class Mobile:
    def __init__(self, brand, model, price):
        self.brand = brand
        self.model = model
        self.price = price

    def display_details(self):
        print(self.brand,self.model,self.price)

obj1 = Mobile("Vivo", "Y27", 12000)
obj2 = Mobile("Vivo", "Y58", 17000)
obj3 = Mobile("iphone", "17 pro", 175000)

obj1.display_details()
obj2.display_details()
obj3.display_details()

```
