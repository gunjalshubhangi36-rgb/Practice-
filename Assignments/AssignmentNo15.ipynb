2. Product Class – __eq__()
Create a Product class with name and price. Implement __eq__() to check whether two
products have the same price.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __eq__(self, other):
        return self.price == other.price

p1 = Product("Laptop", 50000)
p2 = Product("Mobile", 50000)
p3 = Product("Headphones", 3000)

print(p1 == p2)  
print(p1 == p3)
            
        
```

3. Shopping Cart – __len__()
Create a ShoppingCart class containing a list of products. Implement __len__() so that
len(cart) returns the number of products in the cart.

```python
class ShoppingCart:
    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def __len__(self):
        return len(self.products)


cart = ShoppingCart()

cart.add_product("Laptop")
cart.add_product("Mobile")
cart.add_product("Mouse")

print("Number of products:", len(cart))
```

4. Bank Account – __str__()
Create a BankAccount class with account holder name, account number, and balance.
Implement __str__() to display account information when the object is printed.

```python
class BankAccount:
    def __init__(self, name, account_number, balance):
        self.name = name
        self.account_number = account_number
        self.balance = balance

    def __str__(self):
        return f"Account Holder: {self.name}\nAccount Number: {self.account_number}\nBalance: {self.balance}"


account = BankAccount("Shubhangi", "1234567890", 50000)

print(account)
```

5. Number Class – __add__()
Create a Number class that stores one number. Implement __add__() so that two
objects can be added

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value


n1 = Number(10)
n2 = Number(20)

result = n1 + n2

print("Addition:", result)
```

6. Create a Number class and implement:
__add__()
__sub__()
__mul__()
__truediv__()
__mod__()

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

    def __sub__(self, other):
        return self.value - other.value

    def __mul__(self, other):
        return self.value * other.value

    def __truediv__(self, other):
        return self.value / other.value

    def __mod__(self, other):
        return self.value % other.value


n1 = Number(20)
n2 = Number(6)

print("Addition:", n1 + n2)
print("Subtraction:", n1 - n2)
print("Multiplication:", n1 * n2)
print("Division:", n1 / n2)
print("Modulus:", n1 % n2)
```
