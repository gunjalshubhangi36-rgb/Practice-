1. Write a program using a lambda function to find the square of each number in a list
using map(). 

```python
listOne = [6, 5, 2, 3, 4, 1]

listTwo = list(map(lambda num: num*num, listOne))
print(listTwo)
```

2. Write a program using a lambda function to convert all names in a list to uppercase
using map(). 

```python
listOne = ["s", "d", "f", "g", "h", "j"]

listTwo = list(map(lambda i : i.upper(), listOne))
print(listTwo)
```

3. Write a program using map() to calculate a 10% discount on a list of product prices. 


```python
listOne = [60, 87, 87, 90, 45, 23, 12]

listTwo = list(map(lambda i : i*0.10, listOne))
print(listTwo)
```

4. Write a program using filter() to extract all even numbers from a given list. 


```python
listOne = [6, 5, 2, 3, 4, 1]

listTwo = list(filter(lambda num: num%2==0, listOne))
print("all even numbers from a given list: ",listTwo)
               
```

5. Write a program using filter() to display all students who scored 35 or more marks. 


```python
marks = [56, 87, 87, 90, 45, 23, 12]

listTwo = list(filter(lambda num: num>=35, marks))

print("all students who scored 35 or more mark: ",listTwo)
```

6. Write a program using filter() to remove all empty strings from a list. 


```python
listOne = ["Shubhangi", " ", "Divya", " ", "Dipali", " "]

print("original list: ",listOne)

listTwo = list(filter(lambda i : i != " ", listOne))

print("all non-empty strings from a given list: ",listTwo)
```

7. Write a program using map() to find the length of each string in a list. 


```python
listOne = ["Shubhangi", "Divya", "Dipali"]

listTwo = list(map(lambda i : len(i), listOne))
print("length of string: ",listTwo)
```

8. Write a program using reduce() to calculate the sum of all numbers in a list. 


```python
from functools import reduce

listOne = [10, 20, 30, 40, 50]

add = reduce(lambda num1, num2: num1 + num2, listOne)

print(add)

```

9. Given a list of employee salaries, use map() to increase each salary by 15%. 


```python
listOne = [5000, 10000, 15000, 20000]

listTwo = list(map(lambda i : i+i*0.15, listOne))
print("15% of given list: ",listTwo)
```

10. Given a list of integers, use filter() to display only the numbers that are divisible by
both 3 and 5. 

```python
listOne = [6, 5, 2, 3, 4, 1]

listTwo = list(filter(lambda num: num%3==0 or num%5==0, listOne))

print(listTwo)
```
