3. Create an Employee parent class with name and salary. Create a Manager child class
with an additional department attribute. Display all details using constructors.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print("Name of Employee: ", self.name)
        print("Salary of Employee: ", self.salary)

class Manager(Employee):
    def __init__(self, name, salary, address, department):
        super().__init__(name, salary)
        self.address = address
        self.department = department

    def Show(self):
        print("Address :", self.address)
        print("Department: ",self.address)

obj = Manager("Shubhangi", 50000, "Pune", "Comp")
obj.display()
obj.Show()

```

4. Create a BankAccount parent class with account_number and balance. Create a
SavingsAccount child class with an additional interest_rate. Use super() in the child
constructor

```python
class BankAccount:
    def __init__(self, acc_no, balance):
        self.acc_no = acc_no
        self.balance = balance

    def display(self):
        print("Account Number: ", self.acc_no)
        print("Balance: ", self.balance)

class SavingAccount(BankAccount):
    def __init__(self, acc_no, balance, interest_rate):
        super().__init__(acc_no, balance)
        self.interest_rate = interest_rate

    def Show(self):
        print("Interest Rate is: ", self.interest_rate)

obj = SavingAccount(110022, 500000, 1000)
obj.display()
obj.Show()

    
```

Create a Student class with private attributes __name and __marks.
Create getter and setter methods for marks.
Allow marks only between 0 and 100.
Create a child class ExamStudent that inherits from Student

```python
class Student:
    def __init__(self):
        self.__name = "Shubhangi"
        self.__marks = 80

    def get_name(self):
        return self.__name

    def get_marks(self):
   
        if self.__marks >= 0 and self.__marks <= 100:
            return self.__marks
            
        else:
            print("Invalid marks")

        return self.__name

    def set_marks(self, marks):
        self.__marks = marks
      

class ExamStudent(Student):
    def display(self):
        print("name of student: ",self.get_name())
        print("marks of student: ", self.get_marks())

    def update(self, marks):
        self.set_marks(marks)

obj = ExamStudent()

obj.display()

obj.update(70)

obj.display()
```

9. Constructor + Private Data
Create a User class with private __password.
Initialize it through the constructor.
Create get_password() and set_password().
Setter should update the password only if it contains at least 6 characters.

```python
class User:
    def __init__(self):
        self.__password = "123456"

    def get_password(self):
        return self.__password

    def set_password(self, password):
        password = str(password)

        if len(password) >= 6:
            self.__password = password
        else:
            print("Invalid password. Password must have at least 6 characters.")

    def display(self):
        print("Password:", self.get_password())

    def update(self, password):
        self.set_password(password)


obj = User()

obj.display()

obj.update(467776)

obj.display()
```

10. Multilevel Inheritance + Getter/Setter
Person → private age with getter/setter.
Employee → salary.
Manager → department.
Use super() at each level

```python
class Person:
    def __init__(self, age):
        self.__age = age

    def get_age(self):
        return self.__age

    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Invalid age")


class Employee(Person):
    def __init__(self, age, salary):
        super().__init__(age)
        self.salary = salary


class Manager(Employee):
    def __init__(self, age, salary, department):
        super().__init__(age, salary)
        self.department = department


m1 = Manager(30, 60000, "IT")

print("Age:", m1.get_age())
print("Salary:", m1.salary)
print("Department:", m1.department)
```
