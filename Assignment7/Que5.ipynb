5. Write a program to find the missing numbers between 1 to N using sets.
- Input:
- N = 10
- Numbers = {1,2,3,5,7,8,10}
-
- Output:
- {4,6,9}

```python
set1 = {1,2,3,5,7,8,10}
N=10
missing = set()

for i in range(1, 11):
    if i not in set1:
            missing.add(i)
print("missing numbers between 1 to N :",missing)
```
