Python Namespace

there are 4 namespace

1. Built-in Namespace
2. Global Namespace
3. local Namespace
4. Inclosing Namespace

1. Global Namespace
2. local Namespace

```python
name = "mohit"      #global namespace
age = 22


def temp():
    roll = 55       #local namespace

    print("Local namespace: ", roll)         # it can be access only inside the block of code
    print("Global namespace: ",age, name)    # access inside the fun 

print("Global namespace: ",age, name)       # access outside the fun
temp()
```

3. Enclosing Namespace

```python
std_name = "demouser"     #global namespace

def one():
    std_age = 22          #Enclosing namespace
    def two():
        std_roll = 55     #Local namespace
        print(std_roll)
        print(std_age)
        print(std_name)

    two()
one()
```

4. Built-in Namespace

```python
name = "demo"

length = len(name)

print(length)
```

Shadowing

```python
name = "mohit"

def demo():
    name = "demo"        #shadowing
    print(name)
demo()

print(name)

```

global keyword

```python
name = "mohit"

def demo():
    global name
    name = "demo"        #shadowing
    print(name)
demo()

print(name)
```

LEGB  

```python
name = "demo"

def one():
    name = "demo1"
    
    def two():
        name = "demo2"
        print("Local scope: ",name)
    two()
    print("enclosing scope",name)
one()
print("global scope: ",name)
```
