Types of error

1. ZeroDivisionError

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

try:
    res = num1 / num2

except ZeroDivisionError as err:
    print("Error: Cannot divide by zero:", err)

else:
    print("Division of num1 and num2: ", res)

finally:
    print("Division Completed")
```

2. ValueError

```python


try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    res = num1 + num2
    print(res)
except ValueError as err:
    print("Error: please enter valid number: ", err)
else:
    print("Addition of num1 and num2:", res)
```

3. TypeError

```python


try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))
    res = num1 + num2

except TypeError as err :
    print("Error: value is wrong data type: ", err)

else:
    print("Addition of num1 and num2:", res)
    
```

4. IndexError

```python
list1 = [1, 2, 3, 4, 5]

try:
    temp = list1[10]

except IndexError as err:
    print("Error: index out of range: ", err)

else:
    print(temp) 
```

5. KeyError

```python
key = input("Enter key : ")

dict1 = {"name": "Shubhangi", "age": 21, "city": "Pune"}

try:
    temp = dict1[key]

except KeyError as err:
    print("Error: key not found in dictionary: ", err)

else:
    print(temp)
```

6. NameError

```python
num1 = int(input("Enter first number: "))

try:
    res = num1 + num5

except NameError as err:
    print("Error: variable not defined: ", err)

else:
    print(res)
```

7.AttributeError

```python
list1 = [1, 2, 3, 4, 5]

try:
    res = list1.add(6)

except AttributeError as err:
    print("Error: attribute not found: ", err)

else:
    print(res)
```

8. FileNotFoundError

```python


try:
    filedata = open("dem.txt", "r")
    data = filedata.read()

except FileNotFoundError as err:
    print("Error: file not found: ", err)

else:
    print("File data: ", data)
```

9. ImportError

```python

try:
    from math import sqr
    res = sqrt(16)

except ImportError as err:
    print("Error: invalid value for square root: ", err)

else:
    print(res)


```

10. ModuleNoteFound

```python
try:
    import demo

except ModuleNotFoundError as err:
    print("Error: Module are not found: ", err)

    
```

11. Exception

```python
try:
    num1 = int(input("Enter first number: "))
    num2 = input("Enter second number: ")
    res = num1 + num2

except Exception as err :
    print("Error: value is wrong data type: ", err)

else:
    print("Addition of num1 and num2:", res)
```
