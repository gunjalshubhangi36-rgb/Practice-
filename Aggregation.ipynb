Encapsulation

```python
class bank:
    def __init__(self):
        self.name = "demouser"
        self.__pin = 412410

    def getPin(self):
        return self.__pin
    
    def setPin(self,newpin):
        
        strnum = str(newpin)
        if len(strnum) == 6:
            self.__pin = newpin
        else:
            print("Invalid pin")

class subbranch(bank):

    def display(self):
        print(self.name)
        print("Old pin: ",self.getPin())

        self.setPin(123456)

        print("New pin: ",self.getPin())

obj = subbranch()
obj.display()



```
