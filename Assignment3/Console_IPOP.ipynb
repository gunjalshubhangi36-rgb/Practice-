1. Write a program to take a student's name, age, city, and percentage as input and
display them.

```python
name = input("Enter your name: ")
age = input("Enter your age: ")
city = input("Enter your city: ")
percentage = input("Enter your percentage: ")

print("Nmae: ",name)
print("Age: ",age)
print("City: ", city)
print("Percentage: ", percentage)
```

2. Take the length and width of a rectangle as input and calculate:
- Area
- Perimeter

```python
length = float(input("Enter the length of the rectangle: "))
width = float(input("Enter the width of the rectangle: "))

area = length * width
perimeter = 2 * (length + width)


print("Area =", area)
print("Perimeter =", perimeter)
```

3. Take the price and quantity of a product from the user and calculate:
- Total Amount
- GST (18%)
- Final Bill

```python
Price = int(input("Enter Price of product"))
Quantity = int(input("Enter quantity of product"))

totalAmount = Price * Quantity
gst = totalAmount * 0.18
final_bill = totalAmount + gst

print("Total Amount ", totalAmount)
print("GST ", gst)
print("Final Bill ", final_bill)
```

4. Take a student's marks Python, Java, and SQL as input and calculate:
- Total
- Average
- Percentage

```python
Python_Marks = int(input("Enter Python Marks"))
Java_Marks = int(input("Enter java marks"))
SQL_Marks = int(input("Enter SQL Marks"))

Total = Python_Marks + Java_Marks + SQL_Marks
Average = Total / 3
Percentage = (Total/300)*100

print("Total: ",Total)
print("Average: ",Average)
print("Percentage: ", Percentage , "%")
```
