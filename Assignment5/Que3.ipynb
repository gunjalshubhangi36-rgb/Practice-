3. Write a program to remove all duplicate elements from a list while preserving the
original order.
- Input:
- [10,20,30,20,40,10,50]
-
- Output:
- [10,20,30,40,50]


```python
listOne = [10,20,30,20,40,10,50]
result = 0
list=[]
for i in listOne:
    if i not in list:
        list.append(i)

print("Output :",list)
```
