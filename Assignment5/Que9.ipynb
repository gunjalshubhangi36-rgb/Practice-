9. Write a program to find all common elements between two lists.
- List1 = [10,20,30,40]
- List2 = [20,40,60]
-
- Output:
- [20,40]


```python
List1 = [10,20,30,40]
List2 = [20,40,60]

comman = []
for i in List1:
    for i in List2:
        comman.append(i)

print("Comman element in two lists :", comman)
```
