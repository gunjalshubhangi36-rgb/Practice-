10. Write a program to separate all even and odd numbers into two different lists.
- Input:
- [10,15,20,25,30]
-
- Output:
- Even = [10,20,30]
- Odd = [15,25]

```python
listOne = [10,15,20,25,30]
even = []
odd = []

for i in listOne:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)
print("Even :", even)
print("Odd :", odd)
```
