6. Write a program to separate all even and odd elements of a tuple into two different
tuples.
- Input:
- (10,15,20,25,30)
-
- Output:
- Even = (10,20,30)
- Odd = (15,25)

```python
tup1 = (2, 3, 4, 7, 9 ,4 ,6)

even =()
odd =()

for i in tup1:
    if i % 2 == 0:
        even = (i, )
    else:
        odd = (i, )

print("Even Element in tuple :", even)
print("Odd element in tuple :", odd)
```
