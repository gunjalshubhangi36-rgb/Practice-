- b = 4
• Addition
• Subtraction
• Multiplication
• Division

```python
a = 15
b = 4

print("Value of a:", a)
print("Value of b:", b)

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)
```
