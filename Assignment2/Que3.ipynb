3. Create two variables:
- x = "100"
- y = 50
• Convert the variables so that you can perform arithmetic operations between
them.

```python
x = "100"
y = 50

temp1 = int(x)
temp2 = y

print("Value of x is: ", temp1)
print("Value of y is: ", temp2)

print("Addition ", temp1 + temp2)
print("Subtraction ", temp1 - temp2)
print("Multiplication ", temp1 * temp2)
print("Division ", temp1 / temp2)
```
