7. Write a program that converts:
• int → float
• float → string
• string → int
• int → bool
• bool → int


```python
int_no = 20
float_no = float(int_no)
print("Integer->Float:", float_no)

float_no = 20.5
string_no = str(float_no)
print("Float->String:", string_no)

text = "123"
int_no = int(text)
print("String->Integer:", int_no)

int_no = 1
bool_no = bool(int_no)
print("Integer->Boolean:", bool_no)

boll_no = False
int_no = int(boll_no)
print("Boolean->Integer:", int_no)
```
