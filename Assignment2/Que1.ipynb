Create variables of all Python primitive data types. Print their:
• Value
• Data Type
• Memory Address (id())

```python
intrger_num = 50

float_num = 50.5

complex_num = 10 + 5j

bool_value = True

text = "Hello Python"

None_value = None

print('Integer')
print("Value: ",intrger_num)
print("DataType: ",type(intrger_num))
print("Id: ",id(intrger_num))
print()

print("Float")
print("Value: ",float_num)
print("DataType: ",type(float_num))
print("Id: ",id(float_num))
print()


print("Complex")
print("Value: ",complex_num)
print("DataType: ",type(complex_num))
print("Id: ",id(complex_num))
print()


print("Boolen")
print("Value:",bool_value)
print("Datatype: ",type(bool_value))
print("Id: ", id(bool_value))
print()


print("String")
print("Value: ",text)
print("DataType: ",type(text))
print("Id: ",id(text))
print()


print("None")
print("Value: ",None_value)
print("DataType: ",type(None_value))
print("Id: ",id(None_value))
print()


```
