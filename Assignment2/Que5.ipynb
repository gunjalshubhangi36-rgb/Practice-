5. Write a program to calculate electricity bill.
Use variables for:
• Units Consumed
• Cost per Unit
• Calculate total bill and GST (18%).

```python

units_consumed = 250
cost_per_unit = 8

total_bill = units_consumed * cost_per_unit

gst = total_bill * 0.18


final_bill = total_bill + gst

print("Units Consumed:", units_consumed)
print("Cost per Unit: ₹", cost_per_unit)
print("Total Bill (Without GST): ₹", total_bill)
print("GST (18%): ₹", gst)
print("Final Electricity Bill: ₹", final_bill)
```
